```python
!pip install plotly

import plotly.graph_objects as go

# Cidades e coordenadas (longitude, latitude)
locations = {
    "São Paulo": [-46.6333, -23.5505],
    "Recife": [-34.8770, -8.0476],
    "Salvador": [-38.5014, -12.9714],
    "Fortaleza": [-38.5433, -3.7172],
    "Rio de Janeiro": [-43.1729, -22.9068],
    "Belo Horizonte": [-43.9389, -19.9167],
    "Goiânia": [-49.2643, -16.6864],
    "Curitiba": [-49.2731, -25.4284],
    "Porto Alegre": [-51.2300, -30.0346],
    "Uberaba": [-47.9318, -19.7472],
    "Buenos Aires": [-58.4173, -34.6118],
    "Lima": [-77.0428, -12.0464],
    "Santiago do Chile": [-70.6483, -33.4569],
    "Santa Cruz de La Sierra": [-63.1800, -17.7833]
}

# Extrair coordenadas
lons = [locations[city][0] for city in locations]
lats = [locations[city][1] for city in locations]

# Criar gráfico
fig = go.Figure()

fig.add_trace(go.Scattergeo(
    lon = lons,
    lat = lats,
    mode = 'lines+markers+text',
    line = dict(width = 2, color = 'blue'),
    marker = dict(size = 6, color = 'red'),
    text = list(locations.keys()),
    textposition = "top center"
))

# Layout do mapa
fig.update_layout(
    title = 'Rota em forma de rio conectando São Paulo a outras cidades da América Latina',
    geo = dict(
        scope = 'south america',
        projection_type = 'natural earth',
        showland = True,
        landcolor = 'rgb(217, 217, 217)',
        countrycolor = 'rgb(204, 204, 204)',
    ),
    margin = dict(l=0, r=0, t=30, b=0)
)

# Escala de distância (~1000 km)
fig.add_shape(
    type="line",
    x0=0.05, y0=0.05, x1=0.15, y1=0.05,
    xref="paper", yref="paper",
    line=dict(color="black", width=2)
)
fig.add_annotation(
    x=0.10, y=0.02,
    xref="paper", yref="paper",
    text="~1000 km",
    showarrow=False
)

fig.show()
```
